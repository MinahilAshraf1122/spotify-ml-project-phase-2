# **Data Preprocessing**

Check Missing Values

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

Remove Duplicate Rows

```python
df = df.drop_duplicates()
```

Drop Irrelevant Columns

```python
df = df.drop(['Unnamed: 0','track_id'], axis=1)
```

Handle Categorical Features

```python
df = df.drop(['track_name','artists','album_name'], axis=1)
```

Convert Genre into Numeric

```python
df = pd.get_dummies(df, columns=['track_genre'])
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numerical_cols = df.select_dtypes(include=['float64','int64']).columns

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```

Save Preprocessed Dataset

```python
df.to_csv("spotify_preprocessed_dataset.csv", index=False)
```

```python
X = df.drop('popularity', axis=1)
y = df['popularity']
```

Decision Tree model is used to quickly identify important features affecting song popularity. It is computationally efficient and provides feature importance scores.

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(max_depth=5)
model.fit(X, y)

importances = model.feature_importances_

feat_importance = pd.Series(importances, index=X.columns)
feat_importance.sort_values(ascending=False).head(10)
```

```python
feat_importance.sort_values(ascending=False).head(10).plot(kind='bar')
```

Linear Regression model is applied to understand the linear relationship between features and song popularity.

```python
from sklearn.linear_model import LinearRegression

model2 = LinearRegression()
model2.fit(X, y)
```

Gradient Boosting model is used to improve prediction performance by combining multiple weak models into a strong model.

```python
from sklearn.ensemble import GradientBoostingRegressor

model3 = GradientBoostingRegressor(n_estimators=50)
model3.fit(X, y)
```

LightGBM is a fast and efficient gradient boosting model used to identify important features and improve prediction performance.

```python
from lightgbm import LGBMRegressor

model4 = LGBMRegressor()
model4.fit(X, y)
```

# **Feature enginnering**
**first new feature creation**


A new feature 'energy_loudness' is created by combining energy and loudness to capture the overall intensity of a song.

```python
df['energy_loudness'] = df['energy'] * df['loudness']
```

A new feature 'energy_loudness' is created by combining energy and loudness to capture the overall intensity of a song.

```python
df['dance_energy'] = df['danceability'] * df['energy']
```

A new feature 'tempo_energy' is created to combine tempo and energy, which may better represent the overall rhythm intensity of a song.

```python
df['tempo_energy'] = df['tempo'] * df['energy']
```

A new feature 'valence_energy' is created by combining valence and energy to capture the relationship between mood and intensity of a song.

```python
df['valence_energy'] = df['valence'] * df['energy']
```

A new feature 'acoustic_instrumental' is created by combining acousticness and instrumentalness to represent the level of non-vocal and acoustic content in a song.

```python
df['acoustic_instrumental'] = df['acousticness'] * df['instrumentalness']
```

KMeans clustering is applied to group similar songs into clusters, and the cluster label is used as a new feature to capture hidden patterns in the data.

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
df['cluster'] = kmeans.fit_predict(X_scaled)
```
